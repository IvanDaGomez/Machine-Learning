# Python Filtering and Data Extraction

```python
import numpy as np
import pandas as pd
file_path: str = 'python-for-data-science/04.Filtering Data/laptop_price.csv'
```

```python
df_laptops = pd.read_csv(file_path, encoding='latin1')
df_laptops.head()
```

```python
# find != HP in "Company"
not_hp_df = df_laptops[df_laptops['Company'] != 'HP']
not_hp_df.value_counts('Company')
```

# Create a column base on a condition

```python
import numpy as np

new_column = np.where(df_laptops['Price_euros'] > 2000, 'Expensive', 'Cheap')
df_laptops['Price Category'] = new_column
df_laptops
```

## Create a column based on Screen Size

```python
df_laptops['screen_size'] = np.where(df_laptops['Inches'] > 15, 'Big', 'Small' )
df_laptops.head()
```

```python
df_laptops.value_counts('screen_size', normalize=True).round(2) * 100
```

## Filter based on 2 or more conditions



```python
df_laptops[(df_laptops['Company'] == 'Apple') & (df_laptops['Price_euros'] > 2000)]
```

```python
# or
df_laptops[(df_laptops['Company'] == 'Apple') | (df_laptops['Price_euros'] > 2000)]
```

```python
# Create an array based on multiple price tiers(+2 choices)

# Create conditions 
conditions = [
  df_laptops['Price_euros'] > 3000,
  (df_laptops['Price_euros'] > 2000) & (df_laptops['Price_euros'] <=3000),
  (df_laptops['Price_euros'] > 800) & (df_laptops['Price_euros'] <=2000),
  df_laptops['Price_euros'] <= 800
]

values = ['Too Expensive', 'Expensive', 'Affordable', 'Cheap']
conditions
```

```python
df_laptops["Price_Conclution"] = np.select(conditions, values, default='Cheap')
df_laptops.head()
```

```python
# By default .duplicated will only keep the first duplicated element
df_laptops[df_laptops.duplicated('laptop_ID', keep='first')]
```

```python
df_laptops[df_laptops.duplicated(['Product', 'TypeName', 'Inches'])]
```

```python
# ~ is the not operator (returns the not duplicated values)
df_not_typename_duplicated = df_laptops[~df_laptops.duplicated('TypeName', keep='first')]
df_not_typename_duplicated
```

```python
df_not_typename_duplicated[['Company', 'Price_Conclution']]
```
