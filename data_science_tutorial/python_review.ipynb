```python
# Demonstrating list.copy() vs list[:]
original = ['a', 'b', 'c', 'd']

# Method 1: Using list.copy()
copy1 = original.copy()

# Method 2: Using slice notation [:]
copy2 = original[:]

# Both create shallow copies
print("Original:", original)
print("copy1 (using .copy()):", copy1)
print("copy2 (using [:]):", copy2)

# Test if they are the same object (they shouldn't be)
print("\nAre they the same object?")
print("original is copy1:", original is copy1)
print("original is copy2:", original is copy2)
print("copy1 is copy2:", copy1 is copy2)

# Test if they have the same content
print("\nDo they have the same content?")
print("original == copy1:", original == copy1)
print("original == copy2:", original == copy2)
print("copy1 == copy2:", copy1 == copy2)
```

```python
# Testing what happens when we modify the copies
original = ['a', 'b', 'c', 'd']
copy1 = original.copy()
copy2 = original[:]

# Modify the copies
copy1.append('e')
copy2.append('f')

print("After modifications:")
print("Original:", original)  # Should remain unchanged
print("copy1:", copy1)        # Should have 'e' added
print("copy2:", copy2)        # Should have 'f' added

# Both .copy() and [:] create shallow copies, so for simple lists they behave identically
print("\nConclusion: For simple lists, list.copy() and list[:] are functionally equivalent!")
```

## If Statements

```python
# Basic if statement
x = 10
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is equal to 5")
else:
    print("x is less than 5")

# Ternary operator
age = 20
status = "adult" if age >= 18 else "minor"

# Multiple conditions with logical operators
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
    
# Checking if an item exists in a collection
fruits = ["apple", "banana", "cherry"]
if "banana" in fruits:
    print("Yes, banana is in the fruits list")
```

## For Loops

```python
# Basic for loop with a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# For loop with range
for i in range(5):  # 0 to 4
    print(i)

# For loop with range and specified start/step
for i in range(2, 10, 2):  # 2, 4, 6, 8
    print(i)

# For loop with enumeration
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")

# For loop with dictionary
person = {"name": "Ivan", "age": 30, "country": "Colombia"}
for key in person:
    print(f"{key}: {person[key]}")

# Alternative dictionary iteration
for key, value in person.items():
    print(f"{key}: {value}")

# Break statement in a loop
for i in range(10):
    if i == 5:
        break
    print(i)

# Continue statement in a loop
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)  # Only prints odd numbers

# For loop with list comprehension
squares = [x**2 for x in range(1, 6)]  # [1, 4, 9, 16, 25]
```
