## Extracting multiple CSV files by finding a pattern in the URL

https://www.football-data.co.uk/mmz4281/2425/E0.csv

https://www.football-data.co.uk/mmz4281/2425/E1.csv

https://www.football-data.co.uk/mmz4281/2223/E1.csv

```python
def extract_info_from_url(url) -> pd.DataFrame:
  url = 'https://www.football-data.co.uk/mmz4281/2425/E0.csv'
  headers = {
      'User-Agent': 'Mozilla/5.0',
      'Accept': 'text/csv,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
      'Accept-Language': 'en-US,en;q=0.5',
      'Connection': 'keep-alive',
      'Referer': 'https://www.football-data.co.uk/data.php',
      'DNT': '1',
      'Upgrade-Insecure-Requests': '1'
  }

  while True:
    try:
      
        response = requests.get(url, headers=headers)
        df = pd.read_csv(io.StringIO(response.text))

        return df
    
    except requests.exceptions.RequestException as e:

      continue
```

```python
today_year = int(time.strftime('%Y'))
dfs: dict[str, pd.DataFrame] = {}
for i in range(10): # 10 years back
    j = 0
    error_count = 0
    today_year_2_digit = str(today_year - i - 1)[2:]
    last_year_2_digit = str(today_year - i)[2:]
    year_str = f'{today_year_2_digit}{last_year_2_digit}'
    while True:
        url = f'https://www.football-data.co.uk/mmz4281/{year_str}/E{j}.csv'
        extracted_df = extract_info_from_url(url)
        j += 1
        print(extracted_df['ï»¿Div'][0], f'E{j - 1}')
        if extracted_df['ï»¿Div'][0] != f'E{j - 1}':
            print(f'No more data for {year_str} E{j}')
            break
        else:
            dfs[year_str] = extracted_df

        
```

```python
import os
import json
print(type(dfs))
data: dict[str, dict] = {}
for dataframe in dfs.keys():
  dfs[dataframe].fillna(0, inplace=True)
  data[dataframe] = dfs[dataframe].to_dict()

# json.dumps(dfs[dataframe], indent=4)
path: str = 'data/web_scraping_data.json'
with open(path, 'w') as f:
    json.dump(data, f, indent=2)
storage = os.path.getsize(path) / 1e6
print(f'File {path.split('/')[-1]} wrote with {round(storage)} MB')
```

```python
# We can also concat the dfs into a single
df_list:list[pd.DataFrame] = []
for key in dfs.keys():
  df_list.append(dfs[key])
single_df = pd.concat(df_list)
single_df
```
